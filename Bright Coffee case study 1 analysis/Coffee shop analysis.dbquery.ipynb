```sql
-- Running the entire table
SELECT *
FROM workspace.default.coffee_shop;


-- 1. Checking for date ranges
-- The start date for the given data is 2023-01-01 and the end date is 2023-06-30 --
SELECT MIN(transaction_date) AS Start_date,
MAX(transaction_date) AS Last_date
FROM workspace.default.coffee_shop;


-- 2. Checking for different store locations
-- Bright Coffee shop has 3 different locations, namely: Lower Manhattan, Hell's Kitchen and Astoria --
SELECT  DISTINCT store_location
FROM workspace.default.coffee_shop;


-- 3. Checking for different product categories
-- There are 9 different product categories in the dataset --
SELECT DISTINCT product_category
FROM workspace.default.coffee_shop;


-- 3.1. Product_detail
SELECT DISTINCT product_detail
FROM workspace.default.coffee_shop;


-- 3.2. Product_type
SELECT DISTINCT product_type
FROM workspace.default.coffee_shop;


SELECT DISTINCT product_category AS category,
product_detail AS product_name
FROM workspace.default.coffee_shop;


-- 4. Checking for product prices
SELECT MIN(unit_price) AS Cheapest_price
FROM workspace.default.coffee_shop;


-- 4.1. Max price
SELECT MAX(unit_price) AS Highest_price
FROM workspace.default.coffee_shop;


-- 5. Checking for different time_buckets
SELECT DATE_FORMAT(transaction_time, 'HH:mm:ss') AS purchase_time
FROM workspace.default.coffee_shop;


SELECT DISTINCT
Case 
  WHEN DATE_FORMAT(transaction_time, 'HH:mm:ss') BETWEEN '00:00:00' AND '11:59:59' THEN '01. Morning'
  WHEN DATE_FORMAT(transaction_time, 'HH:mm:ss') BETWEEN '12:00:00' AND '16:59:59' THEN '02. Afternoon'
  WHEN DATE_FORMAT(transaction_time, 'HH:mm:ss') BETWEEN '17:00:00' AND '23:59:59' THEN '03. Evening'
  ELSE '04. Night'
END AS time_bucket
FROM workspace.default.coffee_shop;



-- 6. Data aggregating
SELECT COUNT(*) AS number_of_rows,
COUNT(DISTINCT transaction_id) AS number_of_sales,
COUNT(DISTINCT product_id) AS number_of_products,
COUNT(DISTINCT store_id) AS number_of_stores
FROM workspace.default.coffee_shop;


-- 7. DATE Functions
-- 7.1.
SELECT transaction_date,
       Dayname(transaction_date) AS Day_name
FROM workspace.default.coffee_shop;


--7.2. 
SELECT transaction_date,
       Monthname(transaction_date) AS Month_name
FROM workspace.default.coffee_shop;


-- 7.3. Revenue per transaction
SELECT transaction_id,
       transaction_date,
       transaction_qty * unit_price AS revenue_per_transaction
FROM workspace.default.coffee_shop;


-- 7.4. DATE AND TIME, COUNT AND SUM Enhanced data set
SELECT store_location,
       product_category,
       product_detail,
       unit_price,
       transaction_id,
       product_id,
       transaction_qty,
       transaction_date AS purchase_date,
       Dayname(transaction_date) AS Day_name,
       Monthname(transaction_date) AS Month_name,
       Dayofmonth(transaction_date) AS Day_of_month,
       COUNT(DISTINCT transaction_id) AS number_of_sales,
       COUNT(DISTINCT product_id) AS number_of_products,
       SUM(transaction_qty * unit_price) AS revenue_per_day,

       Case
         WHEN Dayname(transaction_date) IN ('Sun', 'Sat') THEN 'Weekend'
         ELSE 'Weekday'
       END AS day_type,
       Case 
         WHEN DATE_FORMAT(transaction_time, 'HH:mm:ss') BETWEEN '00:00:00' AND '11:59:59' THEN '01. Morning'
         WHEN DATE_FORMAT(transaction_time, 'HH:mm:ss') BETWEEN '12:00:00' AND '16:59:59' THEN '02. Afternoon'
         WHEN DATE_FORMAT(transaction_time, 'HH:mm:ss') BETWEEN '17:00:00' AND '23:59:59' THEN '03. Evening'
         ELSE '04. Night'
       END AS time_bucket,
       Case
         WHEN (transaction_qty * unit_price) <=50 THEN '01. Low spend'
         WHEN (transaction_qty * unit_price) BETWEEN 50 AND 100 THEN '02. Medium spend'
         ELSE '03. High spend'
       END AS spend_bucket
FROM workspace.default.coffee_shop
GROUP BY transaction_date,
         Dayname(transaction_date),
         Monthname(transaction_date),
         Dayofmonth(transaction_date),
       Case
          WHEN Dayname(transaction_date) IN ('Sun', 'Sat') THEN 'Weekend'
          ELSE 'Weekday'
         END,
        Case 
          WHEN DATE_FORMAT(transaction_time, 'HH:mm:ss') BETWEEN '00:00:00' AND '11:59:59' THEN '01. Morning'
          WHEN DATE_FORMAT(transaction_time, 'HH:mm:ss') BETWEEN '12:00:00' AND '16:59:59' THEN '02. Afternoon'
          WHEN DATE_FORMAT(transaction_time, 'HH:mm:ss') BETWEEN '17:00:00' AND '23:59:59' THEN '03. Evening'
          ELSE '04. Night'
         END,
         Case
           WHEN (transaction_qty * unit_price) <=50 THEN '01. Low spend'
           WHEN (transaction_qty * unit_price) BETWEEN 50 AND 100 THEN '02. Medium spend'
           ELSE '03. High spend'
         END,
         store_location,
         product_category,
         product_detail,
         unit_price,
         transaction_id,
         product_id,
         transaction_qty;


-- Revenue per day
SELECT transaction_date,
Dayname(transaction_date) AS Day_name,
Monthname(transaction_date) AS Month_name,
Dayofmonth(transaction_date) AS Day_of_month,
COUNT(DISTINCT transaction_id) AS number_of_sales,
SUM(transaction_qty * unit_price) AS revenue_per_day,
COUNT(DISTINCT product_id) AS number_of_products,
store_location

FROM workspace.default.coffee_shop
GROUP BY transaction_date,
         store_location,
         Dayname(transaction_date),
         Monthname(transaction_date),
         Dayofmonth(transaction_date);


-- Revenue per Month and per location
SELECT DISTINCT Monthname(transaction_date) AS Month_name,
COUNT(DISTINCT transaction_id) AS number_of_sales,
SUM(transaction_qty * unit_price) AS revenue_per_month,
COUNT(DISTINCT product_id) AS number_of_products,
store_location
FROM workspace.default.coffee_shop
GROUP BY Monthname(transaction_date),
store_location;
```
